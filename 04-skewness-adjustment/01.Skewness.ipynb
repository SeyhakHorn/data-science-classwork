# Importing Necessary Libraries

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

# Reading Data

```python
women_df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/03.Data_Pattern/01. Skewness Women Health Data.csv')

print('NO OF ROWS AND COLUMN IN DATASET ', women_df.shape,'\n')
print('AFTER READING HEADBRAIN DATASET','\n',women_df.head())
```

# Plot Data using SEABORN KDE (Kernel Density Estimation (KDE) ) PLOT

## It is a statistical visualization method that creates a smooth, continuous curve to represent the probability density distribution of continuous data

```python
sns.histplot(x = women_df['Calcium(mg)'], kde = True, color='c');
```

## CALCULATE AND CHECK SKEWNESS

```python
print(women_df['Calcium(mg)'].skew())

print('GREATER THAN 1 - SO POSITIVELY SKEWED AND HIGHLY SKEWED')
```

# Common transformations include:

## **1. Square Root** (often used with counts data)

## **2. Quarter Root**

## **3. Log (either natural or base 10)**

# APPLY Square Root Method

```python
women_df['Calcium_SquareRoot']=women_df['Calcium(mg)']**(1/2)

women_df
```

## After Squreroot Plot Data using SEABORN KDE (Kernel Density Estimation (KDE) ) PLOT

```python
sns.histplot(x = women_df['Calcium_SquareRoot'], kde = True, color='black');
```

# CHECK SKEWNESS

```python
print(women_df['Calcium_SquareRoot'].skew())

print('SKEWED VALUE IS IN BETWEEN -0.5 to +0.5 -> SO FAIRLY SYMMETRICAL')
```

# Apply Quarter Root

```python
women_df['Calcium_QuarterRoot']=women_df['Calcium(mg)']**(1/4)

women_df
```

```python
sns.histplot(x = women_df['Calcium_QuarterRoot'], kde = True, color='c');
```

## CHECK SKEWNESS

```python
print(women_df['Calcium_QuarterRoot'].skew())

print('SKEWED VALUE IS IN BETWEEN -0.5 to +0.5 -> SO FAIRLY SYMMETRICAL')
```

# Apply Log (either natural or base 10)

```python
women_df['Calcium_LogNatural']=np.log(women_df['Calcium(mg)'])

women_df
```

```python
sns.histplot(x = women_df['Calcium_LogNatural'], kde = True, color='c');
```

## CHECK SKEWNESS

```python
print(women_df['Calcium_LogNatural'].skew())

print('SKEWED VALUE IS IN BETWEEN -1 TO -.5 -> SO NEGATIVE SKEWED AND MODARATE DISTRIBUTION')
```

```python
print(women_df['Iron(mg)'].skew())

print(women_df['Protein(g)'].skew())

print(women_df['Vitamin A(mg)'].skew())

print(women_df['Vitamin C(mg)'].skew())
```
