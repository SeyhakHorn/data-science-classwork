### Task 2 — Profile missing values

```python
print("=== Task 2 — Profile Missing Values ===")
missing_count = missing_value.isnull().sum()
missing_pct = (missing_value.isnull().sum() / len(missing_value)) * 100

missing_df = (
    pd.DataFrame({"Missing Count": missing_count, "Missing Pct (%)": missing_pct})
    .sort_values(by="Missing Pct (%)", ascending=False)
)
print(missing_df)

cols_over_69 = missing_df[missing_df["Missing Pct (%)"] >= 69].index.tolist()
print(f"\nColumns with >= 69% missing data: {cols_over_69}")
```

### Step 3 — Remove fully blank rows

```python
print("=== Step 3 — Remove Fully Blank Rows ===")
fully_blank_mask = missing_value.isnull().all(axis=1)
fully_blank_count = fully_blank_mask.sum()
print(f"a & b) Fully blank rows count: {fully_blank_count}")

# Remove the fully blank rows
df_cleaned = missing_value[~fully_blank_mask].reset_index(drop=True)
print(f"c & d) New shape after removing fully blank rows: {df_cleaned.shape}")
```

### Step 4 — Remove duplicate rows

```python
print("=== Step 4 — Remove Duplicate Rows ===")
duplicate_count = df_cleaned.duplicated().sum()
print(f"a) Duplicate rows count: {duplicate_count}")

df_cleaned = df_cleaned.drop_duplicates().reset_index(drop=True)
print(f"b & c) New shape after removing duplicates: {df_cleaned.shape}")
```

### Step 5 — Recalculate missingness

```python
print("=== Step 5 — Recalculate Missingness ===")
acq_missing_pct = (
    df_cleaned["Acquisition_Channel"].isnull().sum() / len(df_cleaned)
) * 100
print(f"Acquisition_Channel Missing Percentage: {acq_missing_pct:.2f}%")

if acq_missing_pct >= 69:
    df_cleaned = df_cleaned.drop(columns=["Acquisition_Channel"])
    print(
        "Threshold >= 69% met. Dropped 'Acquisition_Channel' from dataframe."
    )
else:
    print("Does not meet >= 69% threshold. Column retained.")
```

### Step 6 — Classify variables

```python
print("=== Step 6 — Classify Variables ===")
classification = """
- Nominal: Gender, City, Payment_Method, Preferred_Support, Favorite_Category, Primary_Device, Employment_Type
- Ordinal: Education_Level, Income_Level, Membership_Tier, Satisfaction_Level
- Discrete Numeric: Total_Orders, Returned_Orders, Monthly_Visits, Support_Tickets
- Continuous Numeric: Avg_Order_Value, Total_Spend, Discount_Percent, Avg_Session_Minutes, Cart_Abandonment_Rate, Avg_Delivery_Days
"""
print(classification)
```

### Step 7 — Nominal-data experiment

```python
print("=== Step 7 — Nominal-Data Experiment ===")
nominal_cols = [
    "Gender",
    "City",
    "Payment_Method",
    "Preferred_Support",
    "Favorite_Category",
]

for col in nominal_cols:
    if col in df_cleaned.columns:
        mode_val = df_cleaned[col].mode()[0]
        missing_cnt = df_cleaned[col].isnull().sum()
        print(f"Column: {col} | Mode: '{mode_val}' | Missing: {missing_cnt}")

        # Choose 'Unknown' category to preserve missingness context
        df_cleaned[col] = df_cleaned[col].fillna("Unknown")

print(
    "\nJustification: Used 'Unknown' category to preserve missingness patterns without artificially inflating mode distributions."
)
```

### Step 8 — Ordinal-data experiment

```python
print("=== Step 8 — Ordinal-Data Experiment ===")
ordinal_cols = [
    "Education_Level",
    "Income_Level",
    "Membership_Tier",
    "Satisfaction_Level",
]

for col in ordinal_cols:
    if col in df_cleaned.columns:
        mode_val = df_cleaned[col].mode()[0]
        print(f"Column: {col} | Mode Category: '{mode_val}'")
        df_cleaned[col] = df_cleaned[col].fillna(mode_val)

print("Ordinal variables imputed using Mode category.")
```

### Step 9 — Discrete-data experiment

```python
print("=== Step 9 — Discrete-Data Experiment ===")
discrete_cols = [
    "Total_Orders",
    "Returned_Orders",
    "Monthly_Visits",
    "Support_Tickets",
]

for col in discrete_cols:
    if col in df_cleaned.columns:
        mean_val = df_cleaned[col].mean()
        median_val = df_cleaned[col].median()
        print(f"Column: {col} | Mean: {mean_val:.2f} | Median: {median_val}")
        # Median chosen to retain integer values
        df_cleaned[col] = df_cleaned[col].fillna(median_val)

print("Discrete count features imputed using Median values.")
```

### Step 10 — Continuous-data experiment

```python
print("=== Step 10 — Continuous-Data Experiment ===")
continuous_cols = [
    "Avg_Order_Value",
    "Total_Spend",
    "Discount_Percent",
    "Avg_Session_Minutes",
    "Cart_Abandonment_Rate",
    "Avg_Delivery_Days",
]

for col in continuous_cols:
    if col in df_cleaned.columns:
        mean_val = df_cleaned[col].mean()
        median_val = df_cleaned[col].median()
        print(f"Column: {col} | Mean: {mean_val:.2f} | Median: {median_val:.2f}")
        # Median chosen for robustness against extreme outliers
        df_cleaned[col] = df_cleaned[col].fillna(median_val)

print("Continuous features imputed using Median values.")
```

### Step 11 — Validate the cleaned dataset




```python
print("=== Step 12 — Validate Cleaned Dataset ===")
print(f"a) Fully blank rows: {df_cleaned.isnull().all(axis=1).sum()}")
print(f"b) Complete duplicate rows: {df_cleaned.duplicated().sum()}")
print(f"c) Total remaining missing values: {df_cleaned.isnull().sum().sum()}")

neg_counts = (
    df_cleaned[
        [c for c in discrete_cols if c in df_cleaned.columns]
    ] < 0
).sum().sum()
print(f"d) Negative counts across discrete columns: {neg_counts}")

pct_cols = [
    c
    for c in ["Discount_Percent", "Cart_Abandonment_Rate"]
    if c in df_cleaned.columns
]
invalid_pcts = (
    (df_cleaned[pct_cols] < 0) | (df_cleaned[pct_cols] > 100)
).sum().sum()
print(f"e) Impossible percentage values: {invalid_pcts}")

print(f"f) Final Dimensions: {df_cleaned.shape[0]} rows, {df_cleaned.shape[1]} columns")
```

##Note on Dropping 'Acquisition_Channel' (Step 5)
Following the rule to drop columns with around 70% missing data, Acquisition_Channel had the highest missing rate at 69.79% (1,347 missing values out of 1,930 remaining rows).Because 69.79% is effectively 70%, the threshold was set to $\ge 69\%$, and Acquisition_Channel was dropped from the dataset.
